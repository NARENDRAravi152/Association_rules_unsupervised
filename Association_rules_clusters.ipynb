# 1 we have to take in uk country and find out this association_rules_products

```python
uk_retail_data = retail_data[retail_data["Country"]=="United Kingdom"]
uk_retail_data
```

# now visualize the pivot table table

```python
uk_pivot_table = pd.pivot_table( data=uk_retail_data,values="Quantity",index="InvoiceNo",columns="Description",aggfunc ="sum").fillna(0)
uk_pivot_table
```

# we have change in one hot encoding .because of apriori ask dataframe in one-hot-encoding formatee

```python
uk_one_hot_pivot_table = uk_pivot_table.applymap(lambda x:0 if x==0 else 1)
uk_one_hot_pivot_table
```

```python
uk_frequent_items = apriori(df = uk_one_hot_pivot_table,min_support=0.02,use_colnames=True)
uk_frequent_items
```

# we can go with association rules function

```python
uk_associate_rules = association_rules(df = uk_frequent_items,metric="confidence",min_threshold=0.02)
uk_associate_rules
```

```python
uk_associate_rules.to_csv("UK_association_data.csv")
```
